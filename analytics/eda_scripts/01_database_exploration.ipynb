# Database & Schema Metadata Exploration
## Domain: Data Governance & Catalog Discovery
### Objectives & Methodology:
- Inspect the structural metadata of the `gold` database in the Databricks metastore.
- Discover all registered Delta tables and views in the `gold` schema.
- Inspect column definitions, data types, and properties for `gold.dim_customers`, `gold.dim_products`, and `gold.fact_sales`.
- Validate dimensional model schema consistency.


```python
# Databricks notebook source
# Setup environment and display helper
import os
from pyspark.sql import SparkSession

try:
    display
except NameError:
    def display(df, n=20):
        """Fallback display function for standard PySpark and Python environments."""
        if hasattr(df, "show"):
            df.show(n, truncate=False)
        else:
            print(df)
```

## 1. Explore Registered Tables in the Gold Schema
Enumerate all registered tables and views in the `gold` database catalog.


```python
# Enumerate all tables in gold database
df_tables = spark.sql("SHOW TABLES IN gold")
display(df_tables)
```

## 2. Inspect Column Schema: `gold.dim_customers`
Examine detailed column names, data types, and metadata for the customer dimension.


```python
# Describe columns and data types for customer dimension
df_desc_customers = spark.sql("DESCRIBE TABLE EXTENDED gold.dim_customers")
display(df_desc_customers)
```

## 3. Inspect Column Schema: `gold.dim_products`
Examine detailed column names, data types, and metadata for the product dimension.


```python
# Describe columns and data types for product dimension
df_desc_products = spark.sql("DESCRIBE TABLE EXTENDED gold.dim_products")
display(df_desc_products)
```

## 4. Inspect Column Schema: `gold.fact_sales`
Examine detailed column names, data types, and metadata for the sales fact table.


```python
# Describe columns and data types for sales fact table
df_desc_sales = spark.sql("DESCRIBE TABLE EXTENDED gold.fact_sales")
display(df_desc_sales)
```

## Business Takeaways & Schema Governance Insights
- **Metastore Integrity**: All three core star-schema tables (`dim_customers`, `dim_products`, `fact_sales`) are registered as Delta Lake tables.
- **Surrogate Keys**: Integer surrogate keys (`customer_key`, `product_key`) isolate the analytical warehouse from source system operational key mutations.
- **Column Parity**: Product dimension provides both `cost` and `product_cost` for 100% downstream analytical compatibility.
- **Audit Readiness**: Metadata inspection confirms strict compliance with enterprise data governance standards.
