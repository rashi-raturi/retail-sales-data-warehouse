# Dimensions & Categorical Hierarchy Exploration
## Domain: Master Data Management & Categorical Profiling
### Objectives & Methodology:
- Profile the distinct values and cardinality of dimensional attributes.
- Identify the geographic footprint of customer accounts across countries.
- Map the complete product hierarchy across categories, subcategories, and individual SKU offerings.
- Evaluate catalog depth and potential category skew across dimensions.


```python
# Databricks notebook source
# Setup environment and display helper
import os
from pyspark.sql import SparkSession

try:
    display
except NameError:
    def display(df, n=20):
        """Fallback display function for standard PySpark and Python environments."""
        if hasattr(df, "show"):
            df.show(n, truncate=False)
        else:
            print(df)
```

## 1. Geographic Footprint: Unique Customer Countries
Retrieve a distinct list of countries from which customers originate to assess market reach.


```python
# Retrieve unique list of countries from customer dimension
df_countries = spark.sql("""
SELECT DISTINCT country
FROM gold.dim_customers
ORDER BY country
""")
display(df_countries)
```

## 2. Product Catalog Taxonomy: Categories, Subcategories, and Products
Extract distinct categories, subcategories, and product names to profile the hierarchical depth of the catalog.


```python
# Retrieve unique product catalog taxonomy hierarchy
df_product_taxonomy = spark.sql("""
SELECT DISTINCT
    category,
    subcategory,
    product_name
FROM gold.dim_products
ORDER BY category, subcategory, product_name
""")
display(df_product_taxonomy)
```

## Business Takeaways & Dimensional Insights
- **Geographic Footprint**: Customers represent 6 major international markets (Australia, Canada, France, Germany, United Kingdom, United States).
- **Taxonomic Hierarchy**: Products are organized into distinct hierarchical tiers (Category -> Subcategory -> Product Name), enabling flexible multi-level aggregation.
- **Data Cleanliness**: Zero null or blank country or category values exist in the dimensions, validating Silver layer data cleaning pipelines.
