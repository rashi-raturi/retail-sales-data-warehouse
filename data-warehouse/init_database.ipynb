# Database and Schema Initialization

## Overview
This Databricks notebook initializes the **Medallion Architecture** environment for the enterprise Data Warehouse.
It sets up the three primary database namespaces (schemas):

- **`bronze`**: Raw data ingestion layer preserving source fidelity from CRM and ERP extracts.
- **`silver`**: Cleaned, standardized, deduplicated, and conformed layer with enterprise audit metadata.
- **`gold`**: Star-schema dimensional model (`dim_customers`, `dim_products`, `fact_sales`) optimized for reporting and analytics.

In Databricks, schemas and databases are synonymous namespaces residing in the Hive Metastore or Unity Catalog.

```python
# Databricks notebook source
# COMMAND ----------
# MAGIC %md
# MAGIC ### Create Medallion Schemas / Databases

# COMMAND ----------
# Create bronze, silver, and gold databases if they do not exist
spark.sql("CREATE DATABASE IF NOT EXISTS bronze")
spark.sql("CREATE DATABASE IF NOT EXISTS silver")
spark.sql("CREATE DATABASE IF NOT EXISTS gold")

print("Databases 'bronze', 'silver', and 'gold' verified/created successfully.")
```

## Environment Verification
Verify that the databases are successfully registered in the catalog/metastore.

```python
# COMMAND ----------
# Helper display function fallback
try:
    display
except NameError:
    def display(df, n=20):
        if hasattr(df, "show"):
            df.show(n, truncate=False)
        else:
            print(df)

# Show registered databases matching the medallion architecture
df_databases = spark.sql("SHOW DATABASES")
medallion_dbs = df_databases.filter("databaseName IN ('bronze', 'silver', 'gold')")
display(medallion_dbs)
```

## Database Metadata Inspection
Inspect the configuration, comments, and storage location for each schema.

```python
# COMMAND ----------
for db_name in ["bronze", "silver", "gold"]:
    print(f"\n================ Database: {db_name} ================")
    db_info = spark.sql(f"DESCRIBE DATABASE EXTENDED {db_name}")
    display(db_info)
```
